```python
numeric_col= [' investment_loss', ' investment_profit', 'age_years', ' weekly_work_hours', ' sample_weight_index', ' education_score']
for i in numeric_col:
    plt.figure(figsize=(6,6))
    sns.histplot(df[i],kde=True,bins=25)
```

```python
for i in numeric_col:
    plt.figure(figsize=(18,6))
    sns.boxplot(x=df[i])
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),annot = True)
```

```python
c=0
for i in df['age_years']:
    if i>122:
        c+=1
print(c)
```

```python
df_cleaner = df[df["age_years"] < 122 ].copy()
```

```python
count=0
for i in df_cleaner['age_years']:
    if i>122:
        count+=1
print(count)
```

```python
categorical_col1 = [' gender_code',' household_role',' annual_income_category', ' demographic_group']
for i in categorical_col1:
    plt.figure(figsize=(8,4))
    sns.countplot(x=df[i])
```
